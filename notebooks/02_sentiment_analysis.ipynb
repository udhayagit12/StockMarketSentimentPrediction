```python
import pandas as pd
from pathlib import Path

# Project path
project_folder = Path(r"D:\Projects\StockMarketSentimentPrediction")

# Load combined dataset
file_path = (
    project_folder
    / "dataset"
    / "processed"
    / "reliance_stock_sentiment.csv"
)

df = pd.read_csv(file_path)

# Convert date
df["Date"] = pd.to_datetime(df["Date"])

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)

# Create next-day closing price
df["Target_Next_Close"] = df["Close"].shift(-1)

# Remove final row because it has no next-day price
df = df.dropna(subset=["Target_Next_Close"]).reset_index(drop=True)

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nSample:")
display(
    df[
        [
            "Date",
            "Close",
            "Daily_Return",
            "MA20",
            "MA50",
            "Volatility",
            "RSI",
            "Daily_Sentiment",
            "Target_Next_Close"
        ]
    ].head(10)
)
```

```python
df["Sentiment_Available"] = (
    df["Daily_Sentiment"].notna().astype(int)
)

print(
    df["Sentiment_Available"].value_counts()
)
```

```python
features = [
    "Close",
    "Volume",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility",
    "RSI",
    "Daily_Sentiment"
]
```

```python
df["Daily_Sentiment"] = df["Daily_Sentiment"].fillna(0)
```

```python
features = [
    "Close",
    "Volume",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility",
    "RSI",
    "Daily_Sentiment"
]

X = df[features]
y = df["Target_Next_Close"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

X_train_scaled = scaler_X.fit_transform(X_train)
X_test_scaled = scaler_X.transform(X_test)

y_train_scaled = scaler_y.fit_transform(
    y_train.values.reshape(-1, 1)
)

y_test_scaled = scaler_y.transform(
    y_test.values.reshape(-1, 1)
)

print("Scaling completed!")
```

```python
import numpy as np

LOOKBACK = 30

X_sequences = []
y_sequences = []

for i in range(LOOKBACK, len(X_train_scaled)):
    X_sequences.append(
        X_train_scaled[i - LOOKBACK:i]
    )
    y_sequences.append(
        y_train_scaled[i]
    )

X_sequences = np.array(X_sequences)
y_sequences = np.array(y_sequences)

print("LSTM input shape:", X_sequences.shape)
print("LSTM target shape:", y_sequences.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

print("TensorFlow version:", tf.__version__)

model = Sequential([
    LSTM(
        64,
        return_sequences=True,
        input_shape=(LOOKBACK, X_sequences.shape[2])
    ),

    Dropout(0.2),

    LSTM(
        32,
        return_sequences=False
    ),

    Dropout(0.2),

    Dense(16, activation="relu"),

    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()
```

```python
history = model.fit(
    X_sequences,
    y_sequences,
    epochs=30,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)

print("LSTM training completed!")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
X_test_sequences = []
y_test_sequences = []

# Combine the final LOOKBACK training observations
# with the test observations so the first test prediction
# has the required historical context.

combined_X = np.concatenate(
    [
        X_train_scaled[-LOOKBACK:],
        X_test_scaled
    ],
    axis=0
)

for i in range(
    LOOKBACK,
    len(combined_X)
):
    X_test_sequences.append(
        combined_X[i - LOOKBACK:i]
    )

X_test_sequences = np.array(X_test_sequences)

y_test_sequences = y_test_scaled.copy()

print("Test sequence shape:", X_test_sequences.shape)
print("Test target shape:", y_test_sequences.shape)
```

```python
predictions_scaled = model.predict(
    X_test_sequences,
    verbose=0
)

print("Predictions generated!")
print("Prediction shape:", predictions_scaled.shape)
```

```python
predictions = scaler_y.inverse_transform(
    predictions_scaled
)

actual_prices = scaler_y.inverse_transform(
    y_test_sequences
)

print("First 5 predictions:")
print(predictions[:5].flatten())

print("\nFirst 5 actual prices:")
print(actual_prices[:5].flatten())
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

actual = actual_prices.flatten()
predicted = predictions.flatten()

mae = mean_absolute_error(
    actual,
    predicted
)

rmse = np.sqrt(
    mean_squared_error(
        actual,
        predicted
    )
)

r2 = r2_score(
    actual,
    predicted
)

print("========== MODEL PERFORMANCE ==========")
print(f"MAE  : ₹{mae:.2f}")
print(f"RMSE : ₹{rmse:.2f}")
print(f"R²   : {r2:.4f}")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# ============================================
# 1. Load processed dataset
# ============================================

project_folder = Path(
    r"D:\Projects\StockMarketSentimentPrediction"
)

file_path = (
    project_folder
    / "dataset"
    / "processed"
    / "reliance_stock_sentiment.csv"
)

df = pd.read_csv(file_path)

df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values("Date").reset_index(drop=True)

# ============================================
# 2. Create next-day target
# ============================================

df["Target_Next_Close"] = df["Close"].shift(-1)

df = df.dropna(
    subset=["Target_Next_Close"]
).reset_index(drop=True)

# ============================================
# 3. Features WITHOUT sentiment
# ============================================

features_without_sentiment = [
    "Close",
    "Volume",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility",
    "RSI"
]

X_base = df[features_without_sentiment]
y_base = df["Target_Next_Close"]

# ============================================
# 4. Chronological train/test split
# ============================================

split_index = int(len(df) * 0.8)

X_base_train = X_base.iloc[:split_index]
X_base_test = X_base.iloc[split_index:]

y_base_train = y_base.iloc[:split_index]
y_base_test = y_base.iloc[split_index:]

print("Training samples:", len(X_base_train))
print("Testing samples:", len(X_base_test))

# ============================================
# 5. Scale data
# ============================================

base_X_scaler = MinMaxScaler()
base_y_scaler = MinMaxScaler()

X_base_train_scaled = base_X_scaler.fit_transform(
    X_base_train
)

X_base_test_scaled = base_X_scaler.transform(
    X_base_test
)

y_base_train_scaled = base_y_scaler.fit_transform(
    y_base_train.values.reshape(-1, 1)
)

y_base_test_scaled = base_y_scaler.transform(
    y_base_test.values.reshape(-1, 1)
)

# ============================================
# 6. Create LSTM sequences
# ============================================

LOOKBACK = 30

X_base_sequences = []
y_base_sequences = []

for i in range(
    LOOKBACK,
    len(X_base_train_scaled)
):
    X_base_sequences.append(
        X_base_train_scaled[
            i - LOOKBACK:i
        ]
    )

    y_base_sequences.append(
        y_base_train_scaled[i]
    )

X_base_sequences = np.array(
    X_base_sequences
)

y_base_sequences = np.array(
    y_base_sequences
)

print(
    "Baseline LSTM input shape:",
    X_base_sequences.shape
)

# ============================================
# 7. Build baseline LSTM
# ============================================

baseline_model = Sequential([
    LSTM(
        64,
        return_sequences=True,
        input_shape=(
            LOOKBACK,
            X_base_sequences.shape[2]
        )
    ),

    Dropout(0.2),

    LSTM(32),

    Dropout(0.2),

    Dense(16, activation="relu"),

    Dense(1)
])

baseline_model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

print("\nBaseline model created!")

baseline_model.summary()
```

```python
baseline_history = baseline_model.fit(
    X_base_sequences,
    y_base_sequences,
    epochs=30,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)

print("Baseline training completed!")
```

```python
# ============================================
# Prepare test sequences
# ============================================

combined_base_X = np.concatenate(
    [
        X_base_train_scaled[-LOOKBACK:],
        X_base_test_scaled
    ],
    axis=0
)

X_base_test_sequences = []

for i in range(
    LOOKBACK,
    len(combined_base_X)
):
    X_base_test_sequences.append(
        combined_base_X[
            i - LOOKBACK:i
        ]
    )

X_base_test_sequences = np.array(
    X_base_test_sequences
)

# ============================================
# Predictions
# ============================================

base_predictions_scaled = baseline_model.predict(
    X_base_test_sequences,
    verbose=0
)

# Convert back to actual ₹ prices
base_predictions = base_y_scaler.inverse_transform(
    base_predictions_scaled
)

base_actual = base_y_scaler.inverse_transform(
    y_base_test_scaled
)

# ============================================
# Calculate metrics
# ============================================

actual = base_actual.flatten()
predicted = base_predictions.flatten()

base_mae = mean_absolute_error(
    actual,
    predicted
)

base_rmse = np.sqrt(
    mean_squared_error(
        actual,
        predicted
    )
)

base_r2 = r2_score(
    actual,
    predicted
)

print("\n========== BASELINE MODEL ==========")
print(f"MAE  : ₹{base_mae:.2f}")
print(f"RMSE : ₹{base_rmse:.2f}")
print(f"R²   : {base_r2:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import load_model

# ============================================
# 1. Load data
# ============================================

project_folder = Path(
    r"D:\Projects\StockMarketSentimentPrediction"
)

data_file = (
    project_folder
    / "dataset"
    / "processed"
    / "reliance_stock_sentiment.csv"
)

df = pd.read_csv(data_file)

df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values("Date").reset_index(drop=True)

# Create next-day target
df["Target_Next_Close"] = df["Close"].shift(-1)

df = df.dropna(
    subset=["Target_Next_Close"]
).reset_index(drop=True)

# ============================================
# 2. Features WITH sentiment
# ============================================

features = [
    "Close",
    "Volume",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility",
    "RSI",
    "Daily_Sentiment"
]

# Missing sentiment = 0 for baseline experiment
df["Daily_Sentiment"] = (
    df["Daily_Sentiment"].fillna(0)
)

X = df[features]
y = df["Target_Next_Close"]

# ============================================
# 3. Train/test split
# ============================================

split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

# ============================================
# 4. Scale
# ============================================

scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

X_train_scaled = scaler_X.fit_transform(X_train)
X_test_scaled = scaler_X.transform(X_test)

y_train_scaled = scaler_y.fit_transform(
    y_train.values.reshape(-1, 1)
)

y_test_scaled = scaler_y.transform(
    y_test.values.reshape(-1, 1)
)

# ============================================
# 5. Create test sequences
# ============================================

LOOKBACK = 30

combined_X = np.concatenate(
    [
        X_train_scaled[-LOOKBACK:],
        X_test_scaled
    ],
    axis=0
)

X_test_sequences = []

for i in range(
    LOOKBACK,
    len(combined_X)
):
    X_test_sequences.append(
        combined_X[i - LOOKBACK:i]
    )

X_test_sequences = np.array(
    X_test_sequences
)

# ============================================
# 6. Load trained sentiment model
# ============================================

model_path = (
    project_folder
    / "models"
    / "reliance_lstm.keras"
)

model = load_model(model_path)

# ============================================
# 7. Predict
# ============================================

predictions_scaled = model.predict(
    X_test_sequences,
    verbose=0
)

predictions = scaler_y.inverse_transform(
    predictions_scaled
)

actual_prices = scaler_y.inverse_transform(
    y_test_scaled
)

# ============================================
# 8. Plot
# ============================================

plt.figure(figsize=(14, 6))

plt.plot(
    actual_prices.flatten(),
    label="Actual Price"
)

plt.plot(
    predictions.flatten(),
    label="Predicted Price"
)

plt.title(
    "RELIANCE: Actual vs Predicted Closing Price"
)

plt.xlabel("Trading Days")
plt.ylabel("Closing Price (₹)")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```
