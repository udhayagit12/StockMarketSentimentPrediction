```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("../dataset/stock_data/reliance.csv")

# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Display basic information
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["Close"])

plt.title("RELIANCE Stock Closing Price")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# Make a copy
df = df.copy()

# Sort by date
df = df.sort_values("Date")

# Remove duplicate rows
df = df.drop_duplicates()

# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove rows with missing values
df = df.dropna()

# Create useful stock features
df["Daily_Return"] = df["Close"].pct_change()

df["MA_7"] = df["Close"].rolling(window=7).mean()
df["MA_21"] = df["Close"].rolling(window=21).mean()

df["Volatility"] = df["Daily_Return"].rolling(window=7).std()

# Remove NaN values created by rolling calculations
df = df.dropna()

print("\nDataset after cleaning:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["Close"], label="Close Price")
plt.plot(df["Date"], df["MA_7"], label="7-Day Moving Average")
plt.plot(df["Date"], df["MA_21"], label="21-Day Moving Average")

plt.title("RELIANCE Stock Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)

plt.show()
```

```python
cleaned_path = "../dataset/stock_data/reliance_cleaned.csv"

df.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print("Saved to:", cleaned_path)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../dataset/stock_data/reliance.csv")

data["Date"] = pd.to_datetime(data["Date"])

print(data.head())
print("\nShape:", data.shape)
```

```python
print("Missing Values:")
print(data.isnull().sum())
```

```python
data["Daily_Return"] = data["Close"].pct_change()

print(data[["Date", "Close", "Daily_Return"]].head(10))
```

```python
data["MA20"] = data["Close"].rolling(window=20).mean()
data["MA50"] = data["Close"].rolling(window=50).mean()

print(data[["Date", "Close", "MA20", "MA50"]].tail())
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    data["Date"],
    data["Close"],
    label="Closing Price"
)

plt.plot(
    data["Date"],
    data["MA20"],
    label="20-Day Moving Average"
)

plt.plot(
    data["Date"],
    data["MA50"],
    label="50-Day Moving Average"
)

plt.title("RELIANCE Stock Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price (₹)")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
data["Volatility"] = data["Daily_Return"].rolling(window=20).std()

print(data[["Date", "Daily_Return", "Volatility"]].tail())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    data["Date"],
    data["Volatility"]
)

plt.title("RELIANCE 20-Day Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    data["Date"],
    data["Volume"]
)

plt.title("RELIANCE Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
delta = data["Close"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()

rs = avg_gain / avg_loss

data["RSI"] = 100 - (100 / (1 + rs))

print(data[["Date", "Close", "RSI"]].tail())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    data["Date"],
    data["RSI"]
)

plt.axhline(70, linestyle="--")
plt.axhline(30, linestyle="--")

plt.title("RELIANCE RSI")
plt.xlabel("Date")
plt.ylabel("RSI")

plt.grid(True)
plt.tight_layout()

plt.show()
```
